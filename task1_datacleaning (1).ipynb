Step 1: Load the data. I loaded the Titanic dataset (891 rows, 12 columns) using Pandas. I checked the data types, missing values, and duplicates.


```python
import pandas as pd

df = pd.read_csv("train.csv")
print("Original shape:", df.shape)
df.info()
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

Step 2: Clean the data.

Duplicates: I checked for duplicate rows and found none.
Age: 177 values were missing. I filled them with the median, because the median is not affected by extreme values.
Embarked: 2 values were missing. I filled them with the most common value (mode).
Cabin: 77% of values were missing, so filling them would be misleading. I removed this column.
Formatting: I changed column names to lowercase, removed extra spaces in text, and changed age to whole numbers.




```python
# Duplicates hatayein
df = df.drop_duplicates()

# Column names lowercase karein
df.columns = df.columns.str.strip().str.lower()

# Missing values
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df = df.drop(columns=["cabin"])

# Formatting
df["name"] = df["name"].str.strip().str.title()
df["sex"] = df["sex"].str.strip().str.lower()
df["age"] = df["age"].round().astype(int)

# Final check
print("Cleaned shape:", df.shape)
print(df.isnull().sum())
df.head()
```

Step 3: Save the data. I saved the cleaned dataset as titanic_cleaned.csv.

```python
df.to_csv("titanic_cleaned.csv", index=False)
```
